```python
import sisl
import matplotlib.pyplot as plt
```

```python
geom1 = sisl.geom.graphene().tile(15,0).tile(15,1)
geom2= sisl.geom.graphene().tile(35,0).tile(35,1)
```

```python
xyz1 = geom1.xyz[:, :2]

fig, ax = plt.subplots()

ax.scatter(*xyz1.T, color="k", s=30, alpha=0.5)

ax.axis("equal")

ax.axis("off")
```

```python
xyz2 = geom2.xyz[:, :2]

fig, ax = plt.subplots()

ax.scatter(*xyz2.T, color="k", s=5, alpha=0.5)

ax.axis("equal")

ax.axis("off")
```
